```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
df.columns
```

```python
# data cleaning: blood pressure --> systolic, diastolic
df['systolic'] = df['blood_pressure'].str.split('/').str[0].astype(int)
df['diastolic'] = df['blood_pressure'].str.split('/').str[1].astype(int)
df = df.drop(columns=['blood_pressure'])
```

```python
df.head()
```

```python
df = df.drop(columns= ['person_id'])
df.head()
```

```python
# EDA
df.duplicated().sum()
```

```python
# Imbalanced multi-class classification
df['sleep_disorder'] = df['sleep_disorder'].fillna('None')
df['sleep_disorder'].value_counts(normalize=True)
```

```python
df['bmi_category'].value_counts(normalize=True)
```

```python
df['bmi_category'] = df['bmi_category'].replace({'Normal Weight': 'Normal'})
df['bmi_category'].value_counts(normalize=True)
```

```python
# Domain features from EDA signals (cardiovascular load features)
df['pulse_pressure'] = df['systolic'] - df['diastolic']
# sleep efficieny/ recovery ration
df['sleep_per_stress'] = (df['sleep_duration'] / (df['stress_level'] + 1)).round(3)

# 4. Group ultra-rare occupations (< 5 rows) to prevent split issues
occ_counts = df['occupation'].value_counts()
rare_occs = occ_counts[occ_counts < 5].index
df['occupation'] = df['occupation'].replace(rare_occs, 'Other')

df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

cols = ['sleep_duration', 'quality_of_sleep', 'systolic', 'diastolic', 'stress_level']
order = ['None', 'Insomnia', 'Sleep Apnea']

fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(15, 8))
axes = axes.flatten()

for i, col in enumerate(cols):
    sns.boxplot(
        data=df,
        x='sleep_disorder',
        y=col,
        order=order,
        palette='Set2',
        ax=axes[i]
    )
    axes[i].set_title(f'{col.replace("_", " ").title()} by Sleep Disorder')
    axes[i].set_xlabel('')

# Hide the unused 6th subplot
axes[-1].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# How strongly Overweight/Obese correlates with Sleep Apnea.
pd.crosstab(df['bmi_category'], df['sleep_disorder'], normalize='index')
```

```python
# How sharply blood pressure and sleep quality separate the three classes
df.groupby('sleep_disorder')[['sleep_duration', 'quality_of_sleep', 'systolic', 'diastolic', 'stress_level', 'pulse_pressure', 'sleep_per_stress']].mean()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction import DictVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, f1_score, roc_auc_score
```

```python
# Stratified split preserves class proportions (~59% None, ~21% Insomnia, ~21% Sleep Apnea)
df_full_train, df_test = train_test_split(
    df, test_size=0.2, random_state=1, stratify=df['sleep_disorder']
)
df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=1, stratify=df_full_train['sleep_disorder']
)

# Encode target strings -> 0, 1, 2
le = LabelEncoder()
y_train = le.fit_transform(df_train['sleep_disorder'])
y_val = le.transform(df_val['sleep_disorder'])
y_test = le.transform(df_test['sleep_disorder'])

# Drop target from feature frames
X_train_df = df_train.drop(columns=['sleep_disorder'])
X_val_df = df_val.drop(columns=['sleep_disorder'])
X_test_df = df_test.drop(columns=['sleep_disorder'])

# Fit DictVectorizer ONLY on train, then transform val and test
dv = DictVectorizer(sparse=False)
X_train = dv.fit_transform(X_train_df.to_dict(orient='records'))
X_val = dv.transform(X_val_df.to_dict(orient='records')) # no fit
X_test = dv.transform(X_test_df.to_dict(orient='records')) # no fit

print(dv.get_feature_names_out())
```

```python
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.tree import export_text

from sklearn.tree import DecisionTreeClassifier, export_text

dt_scores = []
for depth in [2, 3, 4, 5, 6, 8, None]:
    for min_leaf in [1, 3, 5, 10]:
        dt = DecisionTreeClassifier(
            max_depth=depth,
            min_samples_leaf=min_leaf,
            random_state=1
        )
        dt.fit(X_train, y_train)
        y_pred = dt.predict(X_val)
        y_proba = dt.predict_proba(X_val)

        macro_f1 = f1_score(y_val, y_pred, average='macro')
        auc = roc_auc_score(y_val, y_proba, multi_class='ovr')
        dt_scores.append((depth, min_leaf, round(macro_f1, 4), round(auc, 4)))

df_dt_scores = pd.DataFrame(dt_scores, columns=['max_depth', 'min_samples_leaf', 'val_macro_f1', 'val_auc'])
print(df_dt_scores.sort_values(by='val_macro_f1', ascending=False).head(5))

# # Evaluate best Decision Tree on validation set
# best_dt = DecisionTreeClassifier(max_depth=4, min_samples_leaf=3, random_state=1)
# best_dt.fit(X_train, y_train)

# print("\nDecision Tree Validation Report:")
# print(classification_report(y_val, best_dt.predict(X_val), target_names=le.classes_))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_scores = []
for max_depth in [3, 5, 7, 10, None]:
    for min_leaf in [1, 3, 5]:
        for estimator in [50, 100, 200]:
            rf = RandomForestClassifier(
                n_estimators=estimator,
                max_depth=max_depth,
                min_samples_leaf=min_leaf,
                random_state=1,
                n_jobs=-1
            )
            rf.fit(X_train, y_train)
            y_pred = rf.predict(X_val)
            y_proba = rf.predict_proba(X_val)

            macro_f1 = f1_score(y_val, y_pred, average='macro')
            auc = roc_auc_score(y_val, y_proba, multi_class='ovr')
            rf_scores.append((max_depth, min_leaf, estimator, round(macro_f1, 4), round(auc, 4)))

df_rf_scores = pd.DataFrame(
    rf_scores,
    columns=['max_depth', 'min_samples_leaf', 'n_estimators', 'val_macro_f1', 'val_auc']
)
print(df_rf_scores.sort_values(by='val_macro_f1', ascending=False).head(5))

# Evaluate top Random Forest on validation set
best_rf = RandomForestClassifier(n_estimators=100, max_depth=5, min_samples_leaf=3, random_state=1, n_jobs=-1)
best_rf.fit(X_train, y_train)
print("\nRandom Forest Validation Report:")
print(classification_report(y_val, best_rf.predict(X_val), target_names=le.classes_))
```

```python
import xgboost as xgb

features = list(dv.get_feature_names_out())
dtrain = xgb.DMatrix(X_train, label=y_train, feature_names=features)
dval = xgb.DMatrix(X_val, label=y_val, feature_names=features)
dtest = xgb.DMatrix(X_test, label=y_test, feature_names=features)

xgb_scores = []
watchlist = [(dtrain, 'train'), (dval, 'val')]

for eta in [0.03, 0.05, 0.1]:
    for max_depth in [2, 3, 4]:
        for min_child in [1, 3, 5]:
            xgb_params = {
                'eta': eta,
                'max_depth': max_depth,
                'min_child_weight': min_child,
                'subsample': 0.8,
                'colsample_bytree': 0.8,
                'objective': 'multi:softprob',
                'num_class': len(le.classes_),
                'eval_metric': 'mlogloss',
                'seed': 1,
                'verbosity': 0,
            }
            model = xgb.train(
                xgb_params,
                dtrain,
                num_boost_round=200,
                evals=watchlist,
                early_stopping_rounds=15,
                verbose_eval=False
            )
            y_proba = model.predict(dval)
            y_pred = y_proba.argmax(axis=1)

            macro_f1 = f1_score(y_val, y_pred, average='macro')
            auc = roc_auc_score(y_val, y_proba, multi_class='ovr')
            xgb_scores.append((eta, max_depth, min_child, model.best_iteration, round(macro_f1, 4), round(auc, 4)))

df_xgb_scores = pd.DataFrame(
    xgb_scores,
    columns=['eta', 'max_depth', 'min_child_weight', 'best_iter', 'val_macro_f1', 'val_auc']
)
print(df_xgb_scores.sort_values(by='val_macro_f1', ascending=False).head(5))
```

```python
# Finalized model
# 1. Retrain winning Random Forest on combined train + val (df_full_train)
X_full_train = dv.fit_transform(df_full_train.drop(columns=['sleep_disorder']).to_dict(orient='records'))
y_full_train = le.transform(df_full_train['sleep_disorder'])
X_test_final = dv.transform(df_test.drop(columns=['sleep_disorder']).to_dict(orient='records'))

final_rf = RandomForestClassifier(
    n_estimators=50,
    max_depth=3,
    min_samples_leaf=3,
    random_state=1,
    n_jobs=-1
)
final_rf.fit(X_full_train, y_full_train)

# 2. Evaluate ONCE on the held-out test set
y_test_pred = final_rf.predict(X_test_final)
y_test_proba = final_rf.predict_proba(X_test_final)

print("Final Test Set Classification Report:")
print(classification_report(y_test, y_test_pred, target_names=le.classes_))
print("Final Test ROC-AUC (OvR):", round(roc_auc_score(y_test, y_test_proba, multi_class='ovr'), 4))

# 3. Check top feature importances
df_importances = pd.DataFrame({
    'feature': dv.get_feature_names_out(),
    'importance': final_rf.feature_importances_
}).sort_values('importance', ascending=False)

print("\nTop 10 Most Important Features:")
print(df_importances.head(10).to_string(index=False))
```

```python
import json
import pickle

# Load saved model artifact
with open('rf_model.bin', 'rb') as f_in:
    dv, le, model = pickle.load(f_in)

# Mock incoming JSON payload (already cleaned/engineered schema)
incoming_json = '''
{
    "gender": "Female",
    "age": 59,
    "occupation": "Nurse",
    "sleep_duration": 8.1,
    "quality_of_sleep": 9,
    "physical_activity_level": 75,
    "stress_level": 3,
    "bmi_category": "Overweight",
    "heart_rate": 68,
    "daily_steps": 7000,
    "systolic": 140,
    "diastolic": 95,
    "pulse_pressure": 45,
    "sleep_per_stress": 2.025
}
'''

# 3. Parse JSON -> transform with DictVectorizer -> decode back to word label
patient_data = json.loads(incoming_json)
X = dv.transform([patient_data])
pred_idx = model.predict(X)[0]
prediction = le.inverse_transform([pred_idx])[0]

print(prediction)  # Outputs: Sleep Apnea
```
